1. Зібрано дані про співробітників та їхні проєкти: "Ім'я", "Проект", "Години".

2. Створіть DataFrame з не менш ніж вісьмома рядками.

3. Виведіть вихідний DataFrame.

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Ім\'я': ['Анна', 'Борис', 'Анна', 'Галина', 'Галина', 'Борис', 'Борис', 'Захар'],
    'Проект': ['Проект A', 'Проект B', 'Проект B', 'Проект D', 'Проект E', 'Проект D', 'Проект D', 'Проект A'],
    'Години': [120, 150, 100, 130, 140, 160, 110, 170]
})
df
```

4. Підрахуйте загальну кількість годин за кожним співробітником.

```python
hours_by_name = df.groupby('Ім\'я')['Години'].sum()
hours_by_name
```

5. Підрахуйте загальну кількість годин за кожним проєктом.

```python
hours_by_project = df.groupby('Проект')['Години'].sum()
hours_by_project
```

6. Знайдіть співробітника, який витратив найбільше годин.

```python
employee_index = hours_by_name.idxmax()
employee_index
print(employee_index)
max_hours_employee_index = hours_by_name.max()
print(max_hours_employee_index)
print(f"Співробітник, який витратив найбільше годин: {employee_index} з кількістю годин {max_hours_employee_index}")
```
