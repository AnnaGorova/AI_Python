```python
from nltk.corpus import movie_reviews
from nltk.tokenize import word_tokenize
import pandas as pd
```

```python
pos_ids = movie_reviews.fileids("pos")[:10]
neg_ids = movie_reviews.fileids("neg")[:10]
selected_ids = pos_ids + neg_ids
```

```python
rows = []

for fileid in selected_ids:
    text = movie_reviews.raw(fileid)
    tokens = word_tokenize(text)
    category = fileid.split("/")[0]

    rows.append( {
       "fileid": fileid,
       "category": category,
       "token_count": len(tokens),
       "tokens": tokens,
       "first_20_tokens": tokens[:20]
    })

df = pd.DataFrame(rows)
df[["fileid", "category", "token_count", "first_20_tokens"]]
```

```python
total_tokens = df["token_count"].sum()
avg_tokens = df["token_count"].mean()
min_tokens = df["token_count"].min()
max_tokens = df["token_count"].max()

print("Загальна кількість токенів:", total_tokens)
print("Середня кількість токенів:", avg_tokens)
print("Мінімальна довжина:", min_tokens)
print("Максимальна довжина:", max_tokens)
```

### Завдання 2. Класифікатор повідомлень для служби підтримки

Створіть невеликий набір повідомлень користувачів і перевірте, чи можна використати sentiment pipeline як грубий індикатор терміновості.

1. Підготуйте 10 повідомлень клієнтів англійською мовою про різні ситуації:
   - замовлення затрималось;
   - користувач не може увійти в акаунт;
   - оплата пройшла двічі;
   - клієнт дякує за швидку доставку;
   - користувач просто ставить нейтральне питання.
2. Запустіть для них той самий `sentiment-analysis` pipeline.
3. Додайте власне правило:
   - якщо `label == "NEGATIVE"` і `score > 0.90`, поставити `priority = "high"`;
   - інакше поставити `priority = "normal"`.
4. Виведіть таблицю з текстом повідомлення, прогнозом моделі, confidence score і пріоритетом.

Питання для висновку:

- Чи достатньо sentiment analysis для реального визначення пріоритету?
- Які повідомлення модель може оцінити неправильно?

```python
messages = [
    # Замовлення затрималось
    "My order #12345 is delayed for 3 weeks! I need this urgently, this is unacceptable!",
    "I still haven't received my package, it's been 10 days since I ordered. The delivery is very slow.",
    "My order was supposed to arrive yesterday but it didn't. This is extremely frustrating!",
    
    # Проблеми з входом
    "I can't log into my account, it says my password is incorrect but I'm sure it's right.",
    "I forgot my password and the reset email is not arriving. Can someone help me please?",
    "I'm locked out of my account after multiple login attempts. This is very annoying.",
    
    # Проблеми з оплатою
    "My credit card was charged twice for the same order. Please refund the duplicate payment immediately!",
    "I was charged $50 more than the order total. This is a billing error that needs to be fixed.",
    
    # Позитивні повідомлення
    "Thank you for the fast delivery! The package arrived in perfect condition. I love the product!",
    "I just wanted to say that your customer service is excellent. You really helped me solve my problem!",
    
    # Нейтральні питання
    "What are your business hours on weekends? I need to contact support.",
    "Do you ship to international addresses? I want to order from Europe.",
    "Can you tell me more about your return policy? I'm considering buying something."
]

print(f"Підготовлено {len(messages)} повідомлень")
```

```python
results = []
for i, msg in enumerate(messages, 1):
    result = classifier(msg)[0]  
    priority = "HIGH" if result['label'] == "NEGATIVE" and result['score'] > 0.90 else "normal"
    results.append({
        'id': i,
        'message': msg,
        'sentiment': result['label'],
        'confidence': f"{result['score']:.3f}",
        'priority': priority
    })

df_results = pd.DataFrame(results)
```

```python
from tabulate import tabulate

for r in results:
    r['short'] = r['message'][:60] + "..." if len(r['message']) > 60 else r['message']


print(tabulate(results, 
               headers="keys",
               tablefmt='fancy_grid',
               showindex=False))
```
