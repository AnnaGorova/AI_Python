Імпорт бібліотеки для таблиці

```python
import pandas as pd
from tabulate import tabulate
```

Список витрат 


```python
list_expenses = [1200, 1500, 800, 600, 1000, 400, 900]
```

Розоахунки: 
    загальну суму витрат;
    середнє значення;
    максимальна і мінімальна витрата;
    кількість днів, коли витрата була вищою за 1000.

Створення словника та 
вивід таблиці:


```python
total_sum = sum(list_expenses)
print(f"Загальна сума - {total_sum}")
average = total_sum / len(list_expenses)
print(f"Середнє знаення - {round(average, 2)}")
max_expense = max(list_expenses)
print(f"Максимальна витрата - {max_expense}")
min_expense = min(list_expenses)
print(f"Мінімальна витрата - {min_expense}")
days_above_1000 = 0
for x in list_expenses:
    if x > 1000:
        days_above_1000 += 1
print(f"Кількість днів, коли витрата була вищою за 1000 - {days_above_1000}")


data = { 
    "Завдання": [ 
        "Загальна сума", 
        "Середнє значення", 
        "Максимальна витрата", 
        "Мінімальна витрата", 
        "Кількість днів, коли витрата була вищою за 1000" 
    ], 
    "Результат": [ 
        total_sum, 
        round(average, 2),
        max_expense, 
        min_expense, 
        days_above_1000
    ] 
} 
df = pd.DataFrame(data) 
print(tabulate(df, headers='keys', tablefmt='grid'))
```
