5. Пояснити, які ознаки мають потенційно корисний зв'язок із MedHouseVal, а які зв'язки виглядають слабкими або неочевидними.

Серед проведених кореляцій:
* помірний зв'язок відмічено при кореляції MedInc - AveRooms = 0,3 - тобто чим вищий середній дохід, тим більша середян кількість кімнат;
 * потенційно корисний зв'язок виявився у MedInc - MedHouseVal - має середній кореляційний зв'язок = 0,6, що свідчить про те чим вищий  середній дохід, тим вище буде вартість житла. 
 * Досить важливий  виявився зв'язок AveRooms - AveBedrms, який складає 0,8 - сильний кореляційний зв'язок. Напевно, це говорить про покращені умови проживання, кофорт і чітко пов'язуєтьсяя з попереднім показником, що можна описати логічним ланцюгом ==> дохід - платоспроможність - більша кількість кімнат = більша площа (комфорт) - ціни на житло.
    * Всі інші показники мали досить низький кореляційний зв'язок або він взагалі був відсутній - такі ознаки не будуть цінними або корисними.
