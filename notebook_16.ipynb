```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.decomposition import PCA
from sklearn.preprocessing import SplineTransformer
```

```python
data = load_breast_cancer()

X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target, name="target")

X.head()
```

```python
X.describe().T
```

```python
from sklearn.preprocessing import StandardScaler


scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
```

```python
X_scaled_df.describe().T
```

```python
pca_2d = PCA(n_components=2)

X_pca = pca_2d.fit_transform(X_scaled)
```

```python
X_pca_df = pd.DataFrame(X_pca, columns=["PC 1", "PC 2"])
X_pca_df 
```

```python
X_pca_df["target"] = y
X_pca_df['class_name'] = y.map({1: "benign", 0: "malignant"})
X_pca_df
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

for targert_value, class_name in enumerate(data.target_names):
    subset = X_pca_df[X_pca_df["target"] == targert_value]

    ax.scatter(
        subset["PC 1"],
        subset["PC 2"],
        label = class_name,
        alpha = 0.75,
        edgecolors = "black",
        linewidth=0.3

    )

ax.set_title("PCA Projection: BCD")

ax.set_xlabel("PRINCIPAL cOMPONENT 1")
ax.set_ylabel("Principal Component 2")

ax.legend(title="class")

plt.show()
```

```python
axplaied = pd.Series(
    pca_2d.explained_variance_ratio_,
    index=["PC 1", "PC 2"],
    name="exp;ained_variance_ratio"
)
```

```python
print(axplaied.to_frame().round(3))
print(f"Разом перші 2 компоненти пояснюють: {axplaied.sum():.3f} варіативності")
```

```python
pca_full = PCA()
pca_full.fit(X_scaled)


variance_df = pd.DataFrame({
    "component": range(1, len(pca_full.explained_variance_ratio_) + 1),
    "explained": pca_full.explained_variance_ratio_
})


variance_df["cumulative"] = variance_df["explained"].cumsum()


components_for_90 = variance_df.loc[variance_df["cumulative"] >= 0.90, "component"].iloc[0]


print(f" Для 90% дисперсії потрібно {components_for_90} компонент")
print(" Перші 20 компонент:")
print(variance_df.head(20).round(3))
```

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(
    variance_df['component'],
    variance_df['cumulative'],
    marker="o",
    color="navy"
)

ax.axhline(0.90, color="darkred", linestyle="--", label="90%")
ax.axvline(components_for_90, color="darkorange", linestyle="--", label="90% component")

ax.set_title("Cumulative Explained Variance")
ax.set_xlabel("Number of Components")
ax.set_ylabel("Cumulative Explainable Variance")

ax.legend()
plt.show()
```
