```python
import matplotlib.pyplot as plt
```

```python
x = np.arange(1, 10)

y = 5 * x + 10
y
```

```python
from cProfile import label
from tkinter import font

from matplotlib import markers
from matplotlib.lines import lineStyles
from pyparsing import alphas


plt.style.use('default')
plt.figure(figsize = (10, 6))

plt.plot(x, y, color='green', linestyle='--', linewidth = 4, label='line y= 5*x + 10', markersize = 10, marker = 'o')
plt.plot(x, y + 20, color='blue', linestyle=':', linewidth = 4, label='line y= 5*x + 10', markersize = 10, marker = 's')

plt.xlabel('x', fontsize = 12, fontweight = 'bold')
plt.ylabel('y', fontsize = 12, fontweight = 'bold')
plt.title('Line y= 5*x + 10', fontsize = 16, fontweight = 'bold')
plt.grid(alpha=0.3, axis='x')
plt.legend()
plt.show()
```

```python
x1 = np.linspace(1, 10, 1000)
y1 = np.sin(x1)

fig, axes = plt.subplots(figsize=(10, 6))
axes.plot(x1, y1, label='line y= sin(x)')
axes.set_title('Demo line')
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.grid(True)
axes.legend()
```

```python
x2 = np.random.normal(10, 5, 1000)

y2 = np.random.normal(10, 5, 1000)

figure, axes = plt.subplots(2, 2, figsize=(10, 6))
axes[0, 0].scatter(x2, y2, label='Точковий графік', color='purple', alpha=0.5)
axes[0, 1].scatter(y2, x2, label='Точковий графік', color='orange', alpha=0.5)
axes[1, 0].plot(x2, y2, label='Точковий графік', color='green', alpha=0.5)
axes[1, 1].bar(y2, x2, label='Точковий графік', color='red', alpha=0.5)

plt.show()
```

```python
group_a = [170, 175, 180, 185, 190]
group_b = [160, 165, 170, 175, 180]
group_c = [150, 155, 160, 165, 170]

plt.boxplot([group_a, group_b, group_c], tick_labels=['Група A', 'Група B', 'Група C'])
plt.xlabel('Групи')
plt.ylabel('Значення')
plt.title('Boxplot для трьох груп')
plt.show()
```

```python
months = ['Січ', 'Лют', 'Бер', 'Кві', 'Тра', 'Чер']
sales = [150, 200, 250, 300, 280, 320]
plt.figure(figsize=(10, 6))
plt.bar(months, sales, color='blue', edgecolor='black', width=0.6)

plt.xlabel('Місяці', fontsize=12, fontweight='bold')
plt.ylabel('Продажі', fontsize=12, fontweight='bold')
plt.title('Продажі за місяцями', fontsize=14, fontweight='bold')
plt.grid(True)
plt.show()
```
