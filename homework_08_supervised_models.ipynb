```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score
```

```python
california_housing_df = pd.read_csv('https://raw.githubusercontent.com/dmitriy-shtefan/python-ai/main/data/california_housing.csv')
california_housing_df
```

```python
california_housing_df.describe()
```

```python
california_housing_df.describe()['MedHouseVal']
```

```python
X = california_housing_df.drop('MedHouseVal', axis=1)
y = california_housing_df['MedHouseVal']

print(f"X = {X.shape}, y = {y.shape}")
```

```python
median_price = np.median(y)
y_class = (y > median_price).astype(int)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_class, test_size=0.2, random_state=42) 
```

```python
from sklearn.linear_model import LogisticRegression


model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

for i, col in enumerate(X.columns):
    print(f"{col}: {model.coef_[0][i]}")
print()
print(f"intercept = {model.intercept_}\n"
      f"кількість ітерацій = {model.n_iter_[0]}")
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, precision_score, recall_score


cm = confusion_matrix(y_test, y_pred)
print(cm)

accurancy = accuracy_score(y_test, y_pred)
print(f"Частка правильних відповідей серед усіх - {accurancy}") # (TR + TN) / всього   => коли класи збалансовані 0.96

# Зі всіх, кого назвали хворими — скільки справді хворі TP / (TP + FP)
# Коли FP = дорога помилка (спам-фільтр)    => 0.95
precision = precision_score(y_test, y_pred)
print(precision)


# Зі всіх хворих — скільки ми знайшли
# Коли FN = небезпечна помилка (медицина!) => 0.97
recall = recall_score(y_test, y_pred)
print(recall)

# Гармонійне середнє Precision і Recall
# Коли потрібен баланс між P і R  => 0.96

f1 = f1_score(y_test, y_pred)
print(f1)
```
