```python
import matplotlib.pyplot as plt
```

```python
name = ['Alex', 'Irene', 'John', 'Mary', 'Boris']
result_test = [85, 92, 78, 90, 88]

plt.figure(figsize=(10, 6))
plt.bar(name, result_test, label='Результати тестування', color='orange', edgecolor='black', width=0.8)
plt.xlabel('Список імен', fontsize= 14, fontweight='bold')
plt.ylabel('Результат тесту', fontsize=14, fontweight='bold')
plt.title('Результати тестування', fontsize=16, fontweight='bold')
plt.grid(True)
plt.legend()
plt.show()


```
